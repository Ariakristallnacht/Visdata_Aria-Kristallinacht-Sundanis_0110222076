1.   Import Library



```python
import matplotlib.pyplot as plt
```

Dataset

```python
# Membuat data sample
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
product_A_sales = [10, 20, 15, 25, 30, 45, 40, 50, 60, 55, 65, 70]
product_B_sales = [5, 10, 8, 15, 18, 20, 22, 30, 25, 35, 40, 45]
```

Single Line PLot

```python
# line plot 1 garis, style grafik default
plt.plot(months, product_A_sales, label='Product A')
plt.title('Penjualan Poduk A Per Bulan')
plt.xlabel('Bulan')
plt.ylabel('Penjualan')
plt.legend()
plt.grid(True)
plt.show()
```

Multiple Line Plot & Customizations

```python
# Kuatomisasi line plot ( line  1 - Produk A)
plt.plot(months, product_A_sales, label='Product A', color='blue', linestyle='--', marker='o')

# Kuatomisasi line plot ( line  2 - Produk B)
plt.plot(months, product_B_sales, label='Product B', color='red', linestyle='-', marker='o')

plt.title('Penjualan Poduk Per Bulan')
plt.xlabel('Bulan')
plt.ylabel('Jumlah Penjualan')
plt.legend()
plt.grid(True)
plt.show()
```

Jenis Garis untuk Menunjukkan Tren

```python
# Data sample tambahan
product_C_sales = [18, 22, 25, 28, 32, 38, 42, 45, 48, 52, 56, 60]
product_D_sales = [7, 9, 11, 13, 16, 18, 20, 23, 25, 28, 30, 33]

# Plot produk A (dashed)
plt.plot(months, product_A_sales, label='Product A', linestyle='--', color='blue')

# Plot produk B (solid)
plt.plot(months, product_B_sales, label='Product B', linestyle='-', color='red')

# Plot produk C (dash-dot)
plt.plot(months, product_C_sales, label='Product C', linestyle=':', color='green')

# Plot produk D (dotted)
plt.plot(months, product_D_sales, label='Product D', linestyle='-.', color='purple')

plt.title('Penjualan Poduk Per Bulan')
plt.xlabel('Bulan')
plt.ylabel('Jumlah Penjualan')
plt.legend()
plt.grid(True)
plt.show()
```

Sublot pada Line Chart

```python
# Subplot, 2 baris 1 kolom, panjang 10 inch, tinggi 8 inch
fig, axs = plt.subplots(2, 1, figsize=(10, 8))

# Plot pertama untuk produk C
axs[0].plot(months, product_C_sales, label='Product C', linestyle=':', color='green', marker='d')
axs[0].set_title('Penjualan Produk C Per Bulan')
axs[0].set_xlabel('Bulan')
axs[0].set_ylabel('Jumlah Penjualan')
axs[0].legend()
axs[0].grid(True)

# Plot pertama untuk produk D
axs[1].plot(months, product_D_sales, label='Product D', linestyle=':', color='purple', marker='s')
axs[1].set_title('Penjualan Produk D Per Bulan')
axs[1].set_xlabel('Bulan')
axs[1].set_ylabel('Jumlah Penjualan')
axs[1].legend()
axs[1].grid(True)

plt.tight_layout()
plt.show()
```
